## **More about SQL**

Main **categories of SQL commands**


*   Data Definition Language: Modify DB structure (``CREATE``, ``ALTER``, ``DROP``)

*   Data Manipulation Language: Insert, modify, retrieve data (``INSERT INTO``, ``UPDATE``, ``SELECT``, ``DELETE``)

*   Data Control Language: To control access (``GRANT``)


*   Transactional Control Language: To manage database transactions (``ROLLBACK``)


---

**SQL Data Types**


*   `CHAR(n)` fixed length $n$-character string
*   `VARCHAR(n)` variable length string with max size $n$
*   `INT` integer
*   `NUMERIC (x, y)` number with $x$ digits, $y$ of which are after the decimal point
*   `DATE` date value *(year, month, day)*


---

**Syntax Notation**

*   `;` end of the SQL command
*   `Not case sensitive`

## **WHERE**

We might be interested in selecting a specific record, if a condition is satisfied. The ``WHERE`` condition determines which row will be retrieved using the following logical operators.

*   `=` Equal to
*   `<` Less than
*   `>` Greater than
*   `<=` Less than or equal to
*   `>=` Greater than or equal to
*   `!=` Not equal to (or `<>`)

Logical conditions can be combined using `AND` and `OR`

```python
%%sql

-- Retrieve the product ID, product name, vendorID, and product price for each product whose price is above $100
SELECT productid, productname, vendorid, productprice
FROM product
WHERE productprice > 100;
```

```python
%%sql

-- Retrieve the product ID, product name, vendor ID, and product price for each product in the FW category whose price is equal to or below $110

SELECT productid, productname, vendorid, productprice
FROM product
WHERE productprice <= 110 AND categoryid='FW' ;
```

## **HAVING**

Enable summarizations across the groups of related data within tablesd

Determines which groups will be displayed in the result of a query and, consequently, which groups will not be displayed in the results of a query

A query that contains a `HAVING` clause must also contain a `GROUP BY` clause.

```python
%%sql

-- Consider the groups of products where each group contains the products that are from the same category and supplied by the same vendor. For each such group that has more than one product, retrieve the vendorid, product category id, number of product in the group, and average price of the product in the group

SELECT vendorid, categoryid, COUNT(*), AVG(productprice)
FROM product
GROUP BY categoryid, vendorid
HAVING COUNT(*) > 1;
```

```python
%%sql


-- Consider the groups of product where each group contains the products that are from the same category, supplied by the same vendor, and whose product price is $ 50 or higher. For each such group that has more than one product, retrieve the vendor id, product category id, number of products in the group, and average price per product.

SELECT vendorid, categoryid, COUNT(*), AVG(productprice)
FROM product
WHERE productprice >= 50
GROUP BY categoryid, vendorid
HAVING COUNT(*) > 1;
```

```python
%%sql

-- For each product that has more than three items sold within all sales transactions, retrieve the productID value and the total number of product items sold within all sales transaction.

SELECT productid, SUM(noofitems) as total_items
FROM soldvia
GROUP BY productid
HAVING SUM(noofitems) > 3;
```

```python
%%sql


-- For each product that was sold in more than one sales transaction, retrieve the ProductID value and the number of sales transactions in which the product was sold.

SELECT productid, COUNT(tid) as num_transactions
FROM soldvia
GROUP BY productid
HAVING COUNT(tid) > 1;
```

## **OUTER JOIN**

An OUTER JOIN returns all rows from one table and the matched rows from the other table. If there is no match, NULL values are returned for the columns of the non-matching table. There are different types of OUTER JOINs: LEFT (OUTER) JOIN, RIGHT (OUTER) JOIN, and FULL (OUTER) JOIN.

Here's an example of a LEFT (OUTER) JOIN.

```python
%%sql

-- LEFT JOIN example: Get all vendors and the products they supply.
-- If a vendor doesn't supply any products, the product columns will be NULL.
SELECT *
FROM vendor v
LEFT JOIN product p
ON v.vendorid = p.vendorid;
```

RIGHT JOIN

A RIGHT JOIN returns all rows from the right table (the second table in the `FROM` clause) and the matched rows from the left table. If there is no match in the left table, the columns from the left table will have `NULL` values.

*Note: SQLite does not directly support RIGHT JOIN. A common way to achieve the same result in SQLite is to use a LEFT JOIN and reverse the order of the tables.*

Here's how you can achieve the equivalent of a RIGHT JOIN in SQLite, showing all products and their vendors. If a product has no matching vendor (though this shouldn't happen with this schema's foreign keys, it illustrates the concept), the vendor columns would be NULL.

FULL JOIN

A FULL JOIN returns all rows when there is a match in either the left or the right table. If there are rows in one table that do not have matches in the other, those rows will also be included, and the columns from the non-matching table will have `NULL` values.

*Note: SQLite does not directly support FULL JOIN. A common way to achieve the same result in SQLite is to combine a LEFT JOIN and a RIGHT JOIN (simulated with LEFT JOIN by swapping tables) using `UNION` and filtering out duplicates.*
