# **Key Integrity Constraints and Referential Integrity**

Keys and references are the foundations of a relational dataset. If primary keys aren't unique/non-null, rows become indistinguishable. If foreign keys point to missing parents, merges and aggregations silently drop or miscount records. Detecting these early prevents incorrect analysis and highlights data corruptions. Even if your data come from a database **do not take for granted that the process was done correctly**. Always verify.
